```python
import numpy as np

A = np.arange(5, 25).reshape(4, 5)

print("A=\n", A)
print("dtype:", A.dtype)
print("ndim:", A.ndim)
print("shape:", A.shape)
print("size:", A.size)
print("itemsize:", A.itemsize)

B = A.reshape(5, 4)

print("\nB=\n", B)
print("B shape:", B.shape)
```

```python
import numpy as np

a = np.array([[5., 5., 5.],
              [15., 15., 15.],
              [25., 25., 25.]])

b = np.array([2., 4., 6.])

c = a + b

print("c=\n", c)

mask = (c >= 28)

print("\nMask=\n", mask)

count = mask.sum()

print("\nCount:", count)
```

```python
import numpy as np

grades = np.array([[90, 85, 78],
                   [88, 91, 95],
                   [76, 84, 89]])

print("Sum:", np.sum(grades))
print("Min:", np.min(grades))
print("Max:", np.max(grades))
print("Mean:", np.mean(grades))
print("STD:", np.std(grades))
print("VAR:", np.var(grades))

print("Mean per column:", np.mean(grades, axis=0))
print("Mean per row:", np.mean(grades, axis=1))

print("Max per row:", np.max(grades, axis=1))
```

```python
import numpy as np

table = np.array([[5, 8, 2, 7],
                  [1, 9, 4, 3],
                  [6, 0, 11, 10],
                  [12, 13, 14, 15]])

print("Second row:", table[1])

print("Third column:", table[:, 2])

print("Sub table:\n", table[1:3, 1:3])
```

```python
import numpy as np

scores = np.array([[70, 80, 90],
                   [60, 75, 85],
                   [88, 92, 79],
                   [55, 65, 72]])

scores[:, 1] += 5

scores[0] = [100, 100, 100]

scores[:, 2] = 0

print(scores)
```

```python
import numpy as np

A = np.array([[2, 3],
              [4, 5]])

B = np.array([[1, 6],
              [7, 8]])

v = np.array([3, 5])

print("Element-wise:\n", A * B)

print("Matrix multiplication:\n", A @ B)

print("Broadcasting:\n", A * v)
```

```python
import numpy as np

np.random.seed(1)

matrix = np.random.randint(-5, 6, size=(4, 4))

print(matrix)

unique_values, counts = np.unique(matrix, return_counts=True)

print("Unique values:", unique_values)
print("Counts:", counts)

print("Unique rows:\n", np.unique(matrix, axis=0))
```

```python
import numpy as np

x = np.array([4, 2, 7, 1, 9, 5, 3])

A = np.array([[8, 6, 2],
              [1, 7, 4]])

indices = np.argsort(x)

print("Indices:", indices)

print("Sorted x:", x[indices])

min_location = np.unravel_index(A.argmin(), A.shape)

print("Min location:", min_location)

print("Indices > 3:", np.where(x > 3))

print("Split arrays:", np.split(x, [2, 5]))

expanded = np.expand_dims(x, axis=0)

print("Expanded:\n", expanded)

print("Squeezed:", np.squeeze(expanded))
```
