```python
import os

if not os.path.exists("lab4_data"):
    os.mkdir("lab4_data")

for f in ["input", "output", "logs"]:
    path = os.path.join("lab4_data", f)
    if not os.path.exists(path):
        os.mkdir(path)

items = os.listdir("lab4_data")

for item in items:
    path = os.path.join("lab4_data", item)
    if os.path.isfile(path):
        print(item, "File")
    else:
        print(item, "Folder")

os.rename("lab4_data/logs", "lab4_data/history")
```

```python
with open("lab_data/input/notes.txt", "w") as f:
    f.write("Today I studied Python\n")
    f.write("I solved lab 8\n")
    f.write("The tasks were easy\n")
    f.write("I like file handling\n")
    f.write("Next I will study more\n")

with open("lab_data/input/notes.txt", "r") as f:
    lines = f.readlines()
    print(len(lines))

with open("lab_data/output/summary.txt", "w") as out:
    out.writelines(lines)
```
