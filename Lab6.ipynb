```python
students = {}

for i in range(3):
    name = input()
    score = float(input())
    students[name] = score

print(students)
print(max(students, key=students.get))
```

```python
a = set(map(float, input().split()))
b = set(map(float, input().split()))

print(a | b)
print(a & b)
print(a - b)
```
