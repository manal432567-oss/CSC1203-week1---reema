```python
import math

def area(r):
    return math.pi * r * r

circ = lambda r: 2 * math.pi * r

r = float(input())
print(round(area(r), 2))
print(round(circ(r), 2))
```

```python
import re

def sum_numbers(nums):
    return sum(nums)

text = input()
nums = list(map(int, re.findall(r'\d+', text)))
total = sum_numbers(nums)
double = (lambda x: x * 2)(total)

print(nums)
print(total)
print(double)
```

```python
import math

def squares(lst):
    return list(map(lambda x: x*x, lst))

nums = list(map(float, input().split(',')))
sq = squares(nums)

print(sq)
print(max(sq))
print(math.sqrt(sum(sq)))
```

```python
import re
import math

pwd = input()

strong = (re.search(r'[A-Z]', pwd) and
          re.search(r'[a-z]', pwd) and
          re.search(r'\d', pwd))

scores = list(map(float, input().split()))
avg = (lambda x: sum(x)/len(x))(scores)

print("Strong" if strong else "Weak")
print(math.ceil(avg))
```

```python
import re
import math

text = input()

vowel_words = re.findall(r'\b[aeiouAEIOU]\w*', text)
count_vowels = lambda s: len(re.findall(r'[aeiouAEIOU]', s))
lengths = list(map(lambda w: len(w), vowel_words))

avg = round(sum(lengths)/len(lengths), 2) if lengths else 0

print(vowel_words)
print(count_vowels(text))
print(avg)
```
