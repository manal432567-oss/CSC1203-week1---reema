```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Reema", 18)

p.age = 19

print(p.name, p.age)
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


class Manager(Employee):

    def add_bonus(self, bonus):
        self.salary += bonus

    def display(self):
        print(self.name, self.salary)


m = Manager("Reema", 5000)

m.add_bonus(1000)

m.display()
```

```python
class Student:

    def __init__(self):
        try:
            self.name = input()
            self.age = int(input())

            if self.age < 0:
                raise ValueError

        except:
            print("Invalid age")
            self.age = 0
            self.name = "Unknown"

    def display(self):
        print(self.name, self.age)


s = Student()
s.display()
```
