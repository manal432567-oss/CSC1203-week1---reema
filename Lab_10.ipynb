```python
import matplotlib.pyplot as plt

x = [2, 4, 6, 8, 10]
y = [3, 6, 9, 12, 15]

plt.plot(x, y)
plt.title('Sales Growth')
plt.xlabel('Months')
plt.ylabel('Revenue')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 3, 5, 7, 9]
y = [10, 14, 12, 18, 20]

plt.scatter(x, y, s=120, color='red')
plt.title('Student Performance')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 8, 27, 64, 125]

plt.plot(x, y1, label='Linear', marker='o')
plt.plot(x, y2, label='Cubic', linestyle='--', marker='d')

plt.title('Linear vs Cubic')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Bar chart
categories = ['Books', 'Pens', 'Bags']
values = [12, 25, 18]

plt.bar(categories, values)
plt.title('School Items')
plt.xlabel('Items')
plt.ylabel('Quantity')
plt.show()

# Pie chart
langs = ['Python', 'JavaScript', 'Java']
shares = [50, 30, 20]

plt.pie(shares, labels=langs, autopct='%1.1f%%')
plt.title('Programming Languages')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(1, 6)
y1 = x * 3
y2 = x ** 3

plt.figure(figsize=(7, 6))

plt.subplot(2, 1, 1)
plt.plot(x, y1, marker='o')
plt.title('Multiples of 3')
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(x, y2, marker='s')
plt.title('Cubic Function')
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(500)

plt.hist(data, bins=15)
plt.title('Random Distribution')
plt.savefig('random_distribution.png')
plt.show()

plt.boxplot(data)
plt.title('Box Plot')
plt.show()
```
